# Iris Classifier Demo

This notebook demonstrates training, saving, and using the model from `src/`.

```python
from sklearn.datasets import load_iris
from src.model import train_and_evaluate, save_model, load_model

iris = load_iris(as_frame=True)
X = iris.frame[iris.feature_names].to_numpy()
y = iris.frame['target'].to_numpy()

model, metrics = train_and_evaluate(X, y, test_size=0.25, random_state=42, n_estimators=50)
print('Accuracy:', metrics['accuracy'])

save_model(model, 'model.joblib')
```

```python
model = load_model('model.joblib')
sample = X[:5]  # example inputs
print('Sample predictions:', model.predict(sample))
```
